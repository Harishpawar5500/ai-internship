2. List & Dict Comprehensions: Filter even squares into a list and map words to their lengths in a dict.

```python
nums = list(range(1, 11))
even_squares = [n ** 2 for n in nums if n % 2 == 0]

words = ["python", "agent", "vector", "rag"]
word_lengths = {w: len(w) for w in words}

print("Even squares:", even_squares)
print("Word lengths:", word_lengths)
```

3. Functions, Default Arguments, Scope & Docstrings: Write a function with a default parameter and inspect its docstring.

```python
tax_rate = 0.18  # Global scope

def calculate_total(price: float, discount: float = 0.0) -> float:
    """Calculate final price after applying discount and global tax."""
    discounted = price * (1 - discount)  # Local scope
    return round(discounted * (1 + tax_rate), 2)

print("Default discount:", calculate_total(1000))
print("With 10% discount:", calculate_total(1000, discount=0.10))
print("Docstring:", calculate_total.__doc__)
```

5. Refactored Word Counter (3+ Functions with Docstrings): Read file, count normalized words, and display top N frequent words.

```python
import string

def read_file(filepath: str) -> str:
    """Read and return the full text content of a UTF-8 file."""
    with open(filepath, "r", encoding="utf-8") as file:
        return file.read()

def count_words(text: str) -> dict[str, int]:
    """Normalize text, strip punctuation, and return a dictionary of word frequencies."""
    translator = str.maketrans("", "", string.punctuation)
    cleaned_words = text.lower().translate(translator).split()
    counts = {}
    for word in cleaned_words:
        counts[word] = counts.get(word, 0) + 1
    return counts

def get_top_words(word_counts: dict[str, int], top_n: int = 10) -> list[tuple[str, int]]:
    """Return the top_n most frequent (word, count) pairs sorted descending."""
    sorted_items = sorted(word_counts.items(), key=lambda item: item[1], reverse=True)
    return sorted_items[:top_n]

def display_top_words(top_words: list[tuple[str, int]]) -> None:
    """Print formatted ranking of top frequent words."""
    for rank, (word, count) in enumerate(top_words, start=1):
        print(f"{rank}. {word}: {count}")

# Test the pipeline
raw_text = read_file("sample.txt")
frequencies = count_words(raw_text)
top_10 = get_top_words(frequencies, top_n=10)
display_top_words(top_10)
```

```python
script_code = '''import string
import sys


def read_file(filepath: str) -> str:
    """Read and return the full text content of a UTF-8 file."""
    with open(filepath, "r", encoding="utf-8") as file:
        return file.read()


def count_words(text: str) -> dict[str, int]:
    """Normalize text, strip punctuation, and return a dictionary of word frequencies."""
    translator = str.maketrans("", "", string.punctuation)
    cleaned_words = text.lower().translate(translator).split()
    counts = {}
    for word in cleaned_words:
        counts[word] = counts.get(word, 0) + 1
    return counts


def get_top_words(word_counts: dict[str, int], top_n: int = 10) -> list[tuple[str, int]]:
    """Return the top_n most frequent (word, count) pairs sorted descending."""
    sorted_items = sorted(word_counts.items(), key=lambda item: item[1], reverse=True)
    return sorted_items[:top_n]


def main() -> None:
    """Entry point to count and display top 10 words from a given text file."""
    filepath = sys.argv[1] if len(sys.argv) > 1 else input("Enter path to text file: ").strip()
    try:
        text = read_file(filepath)
        counts = count_words(text)
        top_10 = get_top_words(counts, top_n=10)
        print(f"\\nTop {len(top_10)} words in '{filepath}':")
        for rank, (word, count) in enumerate(top_10, start=1):
            print(f"{rank}. {word}: {count}")
    except FileNotFoundError:
        print(f"Error: File '{filepath}' not found.")


if __name__ == "__main__":
    main()
'''

with open("word_count.py", "w", encoding="utf-8") as f:
    f.write(script_code)

print("word_count.py exported successfully!")
```
