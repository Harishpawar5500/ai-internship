1. Rectangle Area & Perimeter: Ask the user for length and width (float) and print area and perimeter.

```python
length = float(input("Enter length: "))
width = float(input("Enter width: "))
area = length * width
perimeter = 2 * (length + width)
print(f"Area: {area:.2f}, Perimeter: {perimeter:.2f}")
```

2. Celsius to Fahrenheit: Convert Celsius input to Fahrenheit formatted to 2 decimal places.

```python
celsius = float(input("Enter temperature in Celsius: "))
fahrenheit = (celsius * 9 / 5) + 32
print(f"{celsius}°C = {fahrenheit:.2f}°F")
```

6. Grade Calculator: Take a score (0-100) and print grade A, B, C, D, or F.

```python
score = float(input("Enter score (0-100): "))
if score < 0 or score > 100:
    print("Invalid score")
elif score >= 90:
    print("Grade: A")
elif score >= 80:
    print("Grade: B")
elif score >= 70:
    print("Grade: C")
elif score >= 60:
    print("Grade: D")
else:
    print("Grade: F")
```

7. Largest of Three Numbers: Ask for three numbers and print the largest using if/elif/else.

8. Leap Year Checker: Take a year and check if it is a leap year.

```python
year = int(input("Enter a year: "))
if (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0):
    print(f"{year} is a Leap Year")
else:
    print(f"{year} is Not a Leap Year")
```

9. Simple Two-Number Calculator: Ask for two numbers and an operator (+, -, *, /) and print result.

```python
num1 = float(input("Enter first number: "))
op = input("Enter operator (+, -, *, /): ")
num2 = float(input("Enter second number: "))

if op == "+":
    print(f"Result: {num1 + num2}")
elif op == "-":
    print(f"Result: {num1 - num2}")
elif op == "*":
    print(f"Result: {num1 * num2}")
elif op == "/":
    if num2 == 0:
        print("Error: Division by zero")
    else:
        print(f"Result: {num1 / num2}")
else:
    print("Invalid operator")
```

14. Vowel and Consonant Counter: Count vowels and consonants in a user-entered string.

```python
text = input("Enter text: ").lower()
vowels = "aeiou"
v_count = 0
c_count = 0

for ch in text:
    if ch.isalpha():
        if ch in vowels:
            v_count += 1
        else:
            c_count += 1

print(f"Vowels: {v_count}, Consonants: {c_count}")
```

15. Factorial Calculator: Compute n! for a non-negative integer n using a loop.

```python
n = int(input("Enter a non-negative integer: "))
fact = 1
for i in range(1, n + 1):
    fact *= i
print(f"{n}! = {fact}")
```

16. Reverse a String: Reverse a user-entered string using a loop.

17. Palindrome Checker: Check if a word is a palindrome (case-insensitive).

```python
word = input("Enter a word: ").strip().lower()
rev = ""
for ch in word:
    rev = ch + rev

if word == rev:
    print("Palindrome")
else:
    print("Not a palindrome")
```

18. Simple Number Guessing Game: Guess a secret number using a while loop with high/low hints.

```python
import random

secret = random.randint(1, 50)
attempts = 0

while True:
    guess = int(input("Guess a number (1-50): "))
    attempts += 1
    if guess < secret:
        print("Too low!")
    elif guess > secret:
        print("Too high!")
    else:
        print(f"Correct! Guessed in {attempts} attempts.")
        break
```

19. Running Total & Average: Keep reading numbers until 'done', then print count, sum, and average.

```python
total = 0.0
count = 0

while True:
    entry = input("Enter a number (or 'done'): ").strip().lower()
    if entry == "done":
        break
    total += float(entry)
    count += 1

if count > 0:
    print(f"Count: {count}, Sum: {total}, Average: {total / count:.2f}")
else:
    print("No numbers entered.")
```

20. Right-Angled Number Triangle: Print a number triangle of n rows.

```python
n = int(input("Enter number of rows: "))
for row in range(1, n + 1):
    for col in range(1, row + 1):
        print(col, end=" ")
    print()
```
